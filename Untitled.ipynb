```python
# GAT
import dgl
import numpy as np
import torch as th
from dgl.nn import GATConv
```

```python
# Case 1: Homogeneous graph
g = dgl.graph(([0,1,2,3,2,5], [1,2,3,4,0,3]))
g = dgl.add_self_loop(g)
feat = th.ones(6, 10)
feat
```

```python
# in_feats, out_feats, num_heads
gatconv = GATConv(10, 2, num_heads=3)
res = gatconv(g, feat)
res
```

```python
# multi-head concatenation
th.reshape(res, (6, 6))
```
